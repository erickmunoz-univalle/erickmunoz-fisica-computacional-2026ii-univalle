**Carpeta y Programa**

```python
%%writefile taller1_file.f90
program taller1_programa_hmax
! Programa para calcular la altura máxima de un proyectil
! Autor: Erick Alexander Muñoz Orozco
! Fecha: 16/09/2026

  implicit none

  ! Definición de constantes y parámetros
  real, parameter:: g = 9.8, pi = 4.0 * atan(1.0) ! Gravedad - Unidad: m/s**2, número pi

  ! Definición de variables de entrada y de cálculo
  real:: velocidad_inicial ! Unidad: m/s
  real:: angulo_grados ! Unidad: grados
  real:: angulo_rad ! Unidad: radianes

  ! Definición de variable de salida
  real:: h_max_calculada ! Altura máxima calculada del proyectil - Unidad: metros

  ! Lectura de entradas del usuario
  print *, "Ingrese el ángulo respecto a la horizontal: "
  read *, angulo_grados
  print *, ""
  print *, "Ingrese la rapidez inicial del proyectil: "
  read *, velocidad_inicial

  ! Conversión de unidades de ángulo en grados a radianes
  angulo_rad = angulo_grados * pi / 180.0

  ! Cálculo matemático de la altura máxima
  h_max_calculada = (velocidad_inicial**2 * (sin(angulo_rad))**2) / (2.0*g)

  ! Impresión de resultados
  print *, ""
  print *, "===================== RESULTADOS ===================="
  print *, "ángulo :", angulo_grados, "grados"
  print *, "rapidez inicial :", velocidad_inicial, "m/s "
  print *, "Altura máxima calculada :", h_max_calculada, "metros"
  print *, "====================================================="
  print *, ""

end program taller1_programa_hmax ! Fin del programa
```

**Casos de Prueba**

```python
%%writefile taller1_pruebas_file.f90
program taller1_pruebas_programa_hmax
! Programa de prueba para calcular la altura máxima de un proyectil
! Autor: Erick Alexander Muñoz Orozco
! Fecha: 16/09/2026

  implicit none

  ! Definición de constantes y parámetros
  real, parameter:: g = 9.8, pi = 4.0 * atan(1.0) ! Gravedad - Unidad: m/s**2, número pi

  ! Definición de variables de entrada y de cálculo
  real:: velocidad_inicial ! Unidad: m/s
  real:: angulo_grados ! Unidad: grados
  real:: angulo_rad ! Unidad: radianes
  real:: h_max_esperada ! Altura máxima esperada - Unidad: metros
  real:: error_relativo ! Unidad: %
  real:: tolerancia_error_relativo = 0.001 ! Tolerancia del error relativo - Unidad: %

  ! Definición de variable de salida
  real:: h_max_calculada ! Altura máxima calculada del proyectil - Unidad: metros

  ! Lectura de entradas del usuario
  print *, "Ingrese el ángulo respecto a la horizontal: "
  read *, angulo_grados
  print *, ""
  print *, "Ingrese la rapidez inicial del proyectil: "
  read *, velocidad_inicial
  print *, ""
  print *, "Ingrese altura máxima esperada"
  read *, h_max_esperada

  ! Conversión de unidades de ángulo en grados a radianes
  angulo_rad = angulo_grados * pi / 180.0

  ! Cálculo matemático de la altura máxima
  h_max_calculada = (velocidad_inicial**2 * (sin(angulo_rad))**2) / (2.0*g)

  ! Cálculo de error relativo
  error_relativo = 100.0 * (abs(h_max_calculada - h_max_esperada) / h_max_esperada)

  ! Impresión de resultados
  print *, ""
  print *, "===================== RESULTADOS ===================="
  print *, "ángulo :", angulo_grados, "grados"
  print *, "rapidez inicial :", velocidad_inicial, "m/s "
  print *, "Altura máxima esperada :", h_max_esperada, "metros"
  print *, "Altura máxima calculada:", h_max_calculada, "metros"
  print *, "Error relativo: ", error_relativo, "%"
  print *, "Tolerancia del error relativo: ", tolerancia_error_relativo, "%"
  if (error_relativo < tolerancia_error_relativo) then
  print *, "Estado del caso: PASS"
  else
  print *, "El caso de prueba ha fallado: FAIL"
  end if
  print *, "====================================================="
  print *, ""

end program taller1_pruebas_programa_hmax ! Fin del programa
```

**Compilación y ejecución de casos de prueba**

```python
!gfortran taller1_pruebas_file.f90 -o taller1_pruebas_programa_hmax && ./taller1_pruebas_programa_hmax
```
