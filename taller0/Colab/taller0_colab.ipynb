```python
# ============================================================
# Colab de prueba — Física Computacional
# Tema: Interés compuesto de una inversión
# ============================================================

# Introducción:
# Este Colab calcula el interés compuesto de una inversión inicial
# de 32 millones de pesos, con una tasa de interés del 9.8% efectivo anual,
# durante un periodo de 6 años.

# ---------------- Programa en Python ----------------
capital = 32_000_000   # pesos
tasa = 0.098           # 9.8% anual
años = 6               # tiempo en años

# Fórmula: M = P * (1 + r)^t
monto_final = capital * (1 + tasa) ** años

# Resultados
print("==============================================")
print("        RESULTADOS DEL CÁLCULO EN PYTHON      ")
print("==============================================")
print("Capital inicial :", capital, "pesos")
print("Tasa de interés :", tasa*100, "% anual")
print("Tiempo :", años, "años")
print("Monto final :", round(monto_final, 2), "pesos")
print("==============================================")
```
